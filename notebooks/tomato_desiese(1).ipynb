now, it is tiome to start with the cnn data preparation. first step is to prepare the data for the cnn


```python
from torchvision import datasets, transforms
```

```python
training_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomVerticalFlip(p=0.5),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5]),
])
evaluation_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5]),
])
```

```python



train_dataset_full_placeholder = datasets.ImageFolder(
    root="/content/tomato-images",
    transform=training_transform
)
evaluation_dataset_full_placeholder = datasets.ImageFolder(
    root="/content/tomato-images",
    transform=evaluation_transform
)
```

the random split results in 3 objects containing the indecies of the training, validation and testing data in the datasets. these indecies can be used to get the subset that we need for each dataset (training, validation and testing).


```python
from torch.utils.data import random_split
import torch

train_size = int(0.8 * len(train_dataset_full_placeholder))
val_size = int(0.1 * len(train_dataset_full_placeholder))
test_size = len(train_dataset_full_placeholder) - train_size - val_size

generator = torch.Generator().manual_seed(42)

training_dataset_indecies_holder, validation_dataset_indecies_holder, testing_dataset_indecies_holder = random_split(
    train_dataset_full_placeholder,
    [train_size, val_size, test_size],
    generator=generator
)
```

time to get the subsets

```python
from torch.utils.data import Subset
```

```python
training_dataset = Subset(train_dataset_full_placeholder, training_dataset_indecies_holder.indices)
validation_dataset = Subset(evaluation_dataset_full_placeholder, validation_dataset_indecies_holder.indices)
testing_dataset = Subset(evaluation_dataset_full_placeholder, testing_dataset_indecies_holder)
```

```python
import os
import shutil

# Define the base directory for the testing data
testing_data_base_path = "/content/testing_data"
os.makedirs(testing_data_base_path, exist_ok=True)

# Get class names from the dataset
class_names = evaluation_dataset_full_placeholder.classes

# Create subfolders for each class within the testing data directory
for class_name in class_names:
    os.makedirs(os.path.join(testing_data_base_path, class_name), exist_ok=True)

# Populate the testing data folders
print("Copying testing images...")
for i, original_idx in enumerate(testing_dataset_indecies_holder.indices):
    # Get original image path and its class label from the full dataset
    image_path, class_label_idx = evaluation_dataset_full_placeholder.imgs[original_idx]
    class_name = class_names[class_label_idx]

    # Construct the destination path
    destination_folder = os.path.join(testing_data_base_path, class_name)
    destination_path = os.path.join(destination_folder, os.path.basename(image_path))

    # Copy the image
    shutil.copy(image_path, destination_path)
    if (i + 1) % 1000 == 0:
        print(f"Copied {i + 1} images.")

print(f"Finished copying {len(testing_dataset_indecies_holder.indices)} testing images.")

# Verify the count in one of the folders as an example
print("\nExample: Listing first 5 images in 'healthy' testing folder:")
if 'healthy' in class_names:
    healthy_test_path = os.path.join(testing_data_base_path, 'healthy')
    print(os.listdir(healthy_test_path)[:5])

print("\nCounts of images in each testing category:")
for category in class_names:
    category_path = os.path.join(testing_data_base_path, category)
    print(f"{category}: {len(os.listdir(category_path))}")
```

now we start with the model

1. the model would have multiple convolutional layers, folowed by pooling layers
2. after the data has passed through these layers, the image gets flattenend into a 1d tensor.
3. then gets inputed to a neural network
4. neural network has multiple hidden layers but 6 output neurons (one for each category of tomatoes)



```python
import torch
import torch.nn as nn
```

```python
class Tomato_classifier(nn.Module):
  def __init__(self):
    super().__init__()


    # creating convolutional layers with nn.Conv2d class and a pool with nn.MaxPool2d class

    # 3 x 224 x 224 in, 32 x 112 x 112 out.
    self.conv1 = nn.Conv2d(3, 32, kernel_size=3, padding=1)
    # 32 x 112 x 112 in, 64 x 56 x 56 out.
    self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
    # 64 x 56 x 56 in, 128 x 28 x 28 out.
    self.conv3 = nn.Conv2d(64, 128, kernel_size=3, padding=1)
    # 128 x 28 x 28 in, 128 x 14 x 14
    self.conv4 = nn.Conv2d(128, 128, kernel_size=3, padding=1)
    # 128 x 14 x 14 in, 128 x 7 x 7
    self.conv5 = nn.Conv2d(128, 128, kernel_size=3, padding=1)
    # n x m x m in, n x m/2 x m/2 out
    self.pool = nn.MaxPool2d(2, 2)


    # creating linear layers

    # 128 x 28 x 28 = 6272 so 6272 inputs an 6 outputs, one for each category
    self.fc1 = nn.Linear(6272, 512)
    self.fc2 = nn.Linear(512, 128)
    self.fc3 = nn.Linear(128, 6)

  def forward(self, x):

      # Convolutional feature extractor
      x = self.pool(torch.relu(self.conv1(x)))
      x = self.pool(torch.relu(self.conv2(x)))
      x = self.pool(torch.relu(self.conv3(x)))
      x = self.pool(torch.relu(self.conv4(x)))
      x = self.pool(torch.relu(self.conv5(x)))

      # Flatten
      x = torch.flatten(x, start_dim=1)

      # Classifier
      x = torch.relu(self.fc1(x))
      x = torch.relu(self.fc2(x))
      x = self.fc3(x)

      return x


```

time to start the training.
1. creating the model and having it on the gpu
2. creating the loss function and the optimizer
3. starting to train

```python
# for drawing the graphs.

import matplotlib.pyplot as plt
from IPython.display import clear_output, display
```

```python
# the points that should be plotted on the graphs

training_losses = []
validation_losses = []
```

```python
def update_plot(
    epoch, epochs, average_train_loss, average_val_loss, validation_accuracy, training_losses, validation_losses, fig, ax
):
    training_losses.append(average_train_loss)
    validation_losses.append(average_val_loss)

    ax.clear()

    ax.plot(training_losses, label="Training Loss")
    ax.plot(validation_losses, label="Validation Loss")

    ax.set_xlabel("Epoch")
    ax.set_ylabel("Loss")
    ax.legend()

    clear_output(wait=True)
    display(fig)

    print(
        f"Epoch {epoch}/{epochs} | "
        f"Train Loss: {average_train_loss:.4f} | "
        f"Val Loss: {average_val_loss:.4f} | "
        f"Val Accuracy: {validation_accuracy:.2f}%"
    )
```

```python

import torch.nn as nn


# creating a models and having it on the GPU

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model1 = Tomato_classifier()

# setting the number of epochs
epochs = 50

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model1.parameters(),
    lr=0.0001
)
```

```python
def train_model(model, criterion, optimizer, epochs, path):



  if os.path.exists(path):
      model.load_state_dict(torch.load(path, map_location=device))
      print("Loaded previous best weights.")

  # for saving the weights of the best pass
  lowest_validation_loss = float("inf")

  # the points that should be plotted on the graphs

  training_losses = []
  validation_losses = []


  # creating the graph
  fig, ax = plt.subplots()


  # mooving the model to the gpu
  model.to(device)
  for epoch in range(epochs):


      # training part
      model.train()

      train_loss = 0

      for batch, labels in training_loader:

          batch = batch.to(device)
          labels = labels.to(device)

          optimizer.zero_grad()

          predictions = model(batch)

          loss = criterion(predictions, labels)

          loss.backward()

          optimizer.step()

          train_loss += loss.item()




      # calculating the avarage training loss
      average_train_loss = train_loss / len(training_loader)


      # validation part no gradient because no weights are being changed
      model.eval()

      val_loss = 0
      correct = 0
      total = 0

      with torch.no_grad():

          for batch, labels in validation_loader:

              batch = batch.to(device)
              labels = labels.to(device)

              predictions = model(batch)

              loss = criterion(predictions, labels)

              val_loss += loss.item()

              predicted = predictions.argmax(dim=1)

              correct += (predicted == labels).sum().item()

              total += labels.size(0)



      # calculating the avarage validation loss and accuracy
      average_val_loss = val_loss / len(validation_loader)
      validation_accuracy = 100 * correct / total

      # plotting

      update_plot(epoch+1, epochs, average_train_loss, average_val_loss, validation_accuracy, training_losses, validation_losses, fig, ax)

      # if the validation loss is lower than the best one, save the weights in the file path

      if(average_val_loss < lowest_validation_loss):
        lowest_validation_loss = average_val_loss
        torch.save({
          "epoch": epoch + 1,
          "model_state": model.state_dict(),
          "optimizer_state": optimizer.state_dict(),
          "validation_loss": average_val_loss,
          "validation_accuracy": validation_accuracy,
        }, path)


  return null


```
